```python

import pandas as pd
import zipfile
from pathlib import Path
from IPython.display import display, Markdown

zip_path = Path("archive.zip")

# Load all CSV files from the ZIP
dataframes = {}

with zipfile.ZipFile(zip_path, "r") as archive:
    for filename in archive.namelist():
        if filename.lower().endswith(".csv"):
            with archive.open(filename) as file:
                dataframes[Path(filename).stem] = pd.read_csv(
                    file,
                    low_memory=False
                )

# Convenient names
respondents = dataframes["respondents"]
ai_trust_ml = dataframes["ai_trust_ml"]
questions = dataframes["questions"]
answer_codes = dataframes["answer_codes"]
multi_select_answers = dataframes["multi_select_answers"]
data_dictionary = dataframes["data_dictionary"]

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 200)

print("All CSV files loaded successfully.")
```

VISUALIZAR LAS TABLAS

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

selector = widgets.Dropdown(
    options=list(dataframes.keys()),
    description="CSV:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

rows = widgets.IntSlider(
    value=20,
    min=5,
    max=100,
    step=5,
    description="Rows:"
)

output = widgets.Output()

def show_table(change=None):
    with output:
        clear_output(wait=True)

        name = selector.value
        df = dataframes[name]

        display(Markdown(f"### {name}.csv"))
        display(df.head(rows.value))

selector.observe(show_table, names="value")
rows.observe(show_table, names="value")

display(widgets.VBox([selector, rows, output]))
show_table()
```
